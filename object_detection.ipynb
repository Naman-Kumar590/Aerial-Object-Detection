```python
from ultralytics import YOLOv10
```

```python
# loading pre trained model 
model = YOLOv10("yolov10b.pt")
model.train(
    data="/home/user/naman/datasets/VisDrone/visdrone.yaml",

    epochs=50,
    freeze=10,

    imgsz=960,
    batch=16,

    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,

    weight_decay=0.0005,

    mosaic=1.0,
    mixup=0.15,

    cos_lr=True,
    patience=50,

    cache=False,
    workers=8,

    project="yolov10_finetuning",
    name="stage1_frozen"
)

```

```python
#Loading new weights
best_stage1_weights = "yolov10_finetuning/stage1_frozen/weights/best.pt"
model_stage2 = YOLOv10(best_stage1_weights)

#Fine tuning the model
model_stage2.train(
    data="/home/user/naman/datasets/VisDrone/visdrone.yaml",
    freeze=0,

    epochs=120,

    imgsz=960,

    batch=8,

    lr0=0.0005,

    optimizer="AdamW",

    mosaic=1.0,

    mixup=0.10,

    cos_lr=True,

    patience=50,

    project="yolov10_finetuning",

    name="stage2_unfrozen"
)
```

```python
import numpy as np
# Patch NumPy 2.0's missing trapz function so torchmetrics doesn't crash
if getattr(np, 'trapz', None) is None:
    np.trapz = np.trapezoid
```

```python
from ultralytics import YOLOv10
model = YOLOv10("yolov10_finetuning/stage2_unfrozen/weights/best.pt")

metrics = model.val(
    data="/home/user/naman/datasets/VisDrone/visdrone.yaml",
    split="val",
    imgsz=960,     
    batch=16,
    conf=0.25      
)

# 3. Print the primary metrics
print("\n--- Validation Results ---")
print(f"mAP50:    {metrics.box.map50:.4f}")
print(f"mAP50-95: {metrics.box.map:.4f}")
print(f"Precision: {metrics.box.mp:.4f}")
print(f"Recall:    {metrics.box.mr:.4f}")
```
